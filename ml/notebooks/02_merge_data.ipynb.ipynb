```python
import pandas as pd

# read CSV, skip first 4 rows
df = pd.read_csv("../data/raw/gdp.csv", skiprows=4)

# check columns again
print(df.columns)
```

```python
# keep only Sri Lanka
df = df[df["Country Name"] == "Sri Lanka"]

# drop extra columns
df = df.drop(columns=["Country Code", "Indicator Name", "Indicator Code"])

# convert to long format
df_long = df.melt(id_vars=["Country Name"], var_name="Year", value_name="GDP")

# rename
df_long = df_long.rename(columns={"Country Name": "Country"})

# remove missing
df_long = df_long.dropna()

# fix year type
df_long["Year"] = df_long["Year"].astype(int)

df_long.head()
```
